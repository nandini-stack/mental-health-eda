#  Mental Health EDA Project
Exploratory Data Analysis on a survey about mental health and related factors.

**Dataset:** Mental Health Dataset from Kaggle

**Goals:**
- Understand demographics
- Investigate relationships between mental health indicators
- Answer key research questions

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set(style='whitegrid')
%matplotlib inline
```

```python
df = pd.read_csv('Mental Health Dataset.csv')
df.head()
```

```python
df.info()

# Check for nulls
df.isnull().sum()
```

##  Gender Distribution

```python
sns.countplot(data=df, x='Gender', order=df['Gender'].value_counts().index)
plt.title('Gender Distribution')
plt.xticks(rotation=45)
plt.show()
```

## Respondents by Country

```python
top_countries = df['Country'].value_counts().head(10)
sns.barplot(y=top_countries.index, x=top_countries.values)
plt.title('Top 10 Countries by Respondents')
plt.xlabel('Number of Respondents')
plt.ylabel('Country')
plt.show()
```

##  Is Stress Related to Mental Health Treatment?

```python
pd.crosstab(df['Growing_Stress'], df['treatment'], normalize='index') * 100
```

##  Mood Swings vs Coping Struggles

```python
pd.crosstab(df['Mood_Swings'], df['Coping_Struggles'], normalize='index') * 100
```

##  Family History vs Work & Habit Changes

```python
pd.crosstab(df['family_history'], df['Changes_Habits'], normalize='index') * 100
```
